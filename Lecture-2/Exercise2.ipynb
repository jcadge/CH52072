# **Lecture 2 - Exercise 2**

For this excerise, you will generate electronic and steric features for a set of carboxylic acid anions in the context of an amide coupling reaction.

To generate the features, we are going to use the AQME tool in the easyROB software.

### Task 1: Generate features from the truncated file

1. First, open easyROB.
2. Select the truncated CSV file, `acid_anions_canonical_truncated.csv`. Note that you'll have to download this file from the GitHub.

3. Enable the AQME Workflow on the main easyROB screen.

4. At the bottom of the window, under QDESCP Atoms, type the SMARTS string for a generic carboxylic acid - `[*]C(O)=O`. And change the Descriptor Level to "full".

5. Back on the main easyROB screen, click "Run AQME (descriptor generation)" to start running the descriptor generation.

In the background this does the following:
- Converts the SMILES strings in the CSV file
- Generates conformers and does conformer sampling
- Runs electronic structure calculations (using the SQM xTB method)
- Collects descriptors
