In this Exercise, you will have a go at clustering the PCA chemical space of the carboxylic acid anions where their properties were calculated using DFT.

First, let's load/install the packages we will need (**you may have to add some more**):

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import molplotly
```

Let's also load in the data, plot it and view the clusters:

```python
df = pd.read_csv("dft_acid_anions_dft_pca-clustered.csv", index_col=0)
df.head()
```

```python
dim1 = "PC1"
dim2 = "PC2"

plt.figure(figsize=(8, 6))
plt.scatter(df[dim1], df[dim2], c=df["Cluster"], cmap="tab20", alpha=0.7)
plt.xlabel(dim1)
plt.ylabel(dim2)
plt.show()
```

```python
fig = px.scatter(df, x=dim1, y=dim2, color="Cluster", hover_data=["SMILES String"], width=1200, height=800, color_continuous_scale=px.colors.qualitative.T10)

app = molplotly.add_molecules(fig=fig,
                              df=df,
                              smiles_col="SMILES String",
)

app.run(mode='inline', port=8700, height=1000)
```
